## Criação de novas colunas

```python
#criando as categorias a partir do nps_score
# aparentemente np.select() é mais eficiente
dados["nps_categoria"] = dados["nps_score"].case_when([
    (dados["nps_score"] >= 9, "Promotores"),
    ((dados["nps_score"] >= 7) & (dados["nps_score"] < 9), "Neutros"),
    ((dados["nps_score"] >= 0) & (dados["nps_score"] < 7), "Detratores")
    ])


dados[["nps_score", "nps_categoria", "csat_internal_score"]].head(20)
```

```python
#variável categórica se teve uma ou mais reclamação ou não
dados["complaints"] = np.where(dados["complaints_count"] > 0, 1, 0)

dados[["complaints", "complaints_count"]].head(20)
```
